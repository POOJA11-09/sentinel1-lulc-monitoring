```python
import shutil
from importlib import metadata

packages = ["numpy", "pandas", "rasterio", "GDAL", "pyproj", "shapely", "geopandas",
            "xarray", "rioxarray", "matplotlib", "scikit-learn", "scipy", "tqdm",
            "torch", "torchvision", "segmentation-models-pytorch", "timm", "albumentations",
            "tensorflow", "earthengine-api", "gdown", "google-api-python-client", "requests"]

installed = {}
for p in packages:
    try:
        installed[p] = metadata.version(p)
    except metadata.PackageNotFoundError:
        installed[p] = None
    print(f"{p:32s} {installed[p] or '--- NOT INSTALLED ---'}")

print("\nCommand-line tools:")
for tool in ["nvidia-smi", "gdalinfo", "rclone", "gdown", "git", "unzip"]:
    print(f"  {tool:12s} {shutil.which(tool) or '--- not found ---'}")

print("\nDisk space:")
for label, path in [("vault (persistent)", "/home/jovyan/vault"),
                    ("home (temporary)", "/home/jovyan"),
                    ("tmp (temporary)", "/tmp")]:
    total, used, free = shutil.disk_usage(path)
    print(f"  {label:20s} total {total/1e9:8.1f} GB | free {free/1e9:8.1f} GB")
```

```python
import os
try:
    import torch
    print("PyTorch version :", torch.__version__)
    print("CUDA available  :", torch.cuda.is_available())
    print("CUDA (PyTorch)  :", torch.version.cuda)
    print("cuDNN version   :", torch.backends.cudnn.version())
    print("GPU count       :", torch.cuda.device_count())
    for i in range(torch.cuda.device_count()):
        prop = torch.cuda.get_device_properties(i)
        print(f"GPU {i}           : {prop.name} | VRAM {prop.total_memory/1024**3:.1f} GB")
    if torch.cuda.is_available():
        x = torch.randn(2000, 2000, device="cuda")
        print("GPU test        : matrix multiply OK ->", round(float((x @ x).mean()), 4))
except ImportError:
    print("PyTorch is NOT installed.")

print()
print(os.popen("nvidia-smi --query-gpu=name,memory.total,memory.used,driver_version --format=csv").read())
```

```python
import urllib.request

for url in ["https://www.google.com", "https://drive.google.com",
            "https://www.googleapis.com", "https://pypi.org"]:
    try:
        code = urllib.request.urlopen(url, timeout=10).status
        print(f"{url:30s} reachable (HTTP {code})")
    except Exception as e:
        print(f"{url:30s} NOT reachable -> {type(e).__name__}: {e}")
```
